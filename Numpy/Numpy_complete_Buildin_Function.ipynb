#   1.Array Creation Functions

```python
import numpy as np
```

```python
a = np.array([1,2,3])
print("Array a:",a)
```

```python
# Create an array with evenly spaced values 
b = np.arange(0,10,2) # Values from 0 to 10 with 2 step
print("Array b:",b)
```

```python
d = np.zeros([2,3])
print('Array d:\n',d)
```

```python
e = np.ones ([3,2],dtype = int)
print('Array e:\n',e)
```

```python
# Create an identity matrix
f = np.eye(4)
print("Identity matrix f:\n",f)
```

```python
f = np.eye(3,4,k=1)  # k is index of the diagonal (0 = main diagonal, positive = above, negative = below)
print("Identity matrix f:\n",f)
```

# 2.Array Manipulation Functions

```python
# Reshape an array

a1 = np.array([1,2,3])
reshaped = np.reshape(a1,(3,1))
print(reshaped)
```

```python
# Flatten an array

f1 = np.array([[1,2,3],[3,4,5],[5,6,7]])
flattened = np.ravel(f1)
print(flattened)
```

```python
# Transpose an array
e1 = np.array([[1,2],[3,4]])
transpose = np.transpose(e1)
print(transpose)
```

```python
# stack arrays vertically (vstack)
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.vstack([a2,b2])
print(stacked)
```

```python
# stack arrays horizontally (hstack)
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.hstack([a2,b2])
print(stacked)
```

# 3.Mathematical Functions

```python
# Add two arrays
g = np.array([1,2,3,4])
added = np.add(g,2)
print('add 2 to g',added)
```

```python
# square each element
g = np.array([1,2,3,4])
squared = np.power(g,2)
print('Squared g:',squared)
```

```python
# Square root of each element
sqrt_val = np.sqrt(g)
print('square root of g:',sqrt_val)
```

```python
# Dot- Product
a = np.array([1,2,3])
b = np.array([6,5,9])
dot_product = np.dot(a,b)
print(dot_product)
```

```python
# Matrix multiplication
np.random.seed(1)   # it is fix the random generator elements
a = np.random.randint(1,5,[2,2])
b = np.random.randint(1,5,[2,2])
print(a)
print(b)
c= np.dot(a,b)
print("Dot product:\n",c)
```

# 4.Statistical Functions

```python
# Mean
s = np.array([1,2,3,4])
mean = np.mean(s)
print(mean)
```

```python
# variance 
s = np.array([1,2,3,4])
variance = np.var(s)
print(variance)
```

```python
# Standard deviation
s = np.array([1,2,3,4])
std_dev = np.std(s)
print(std_dev)
```

```python
# Minimum and Maximum
s = np.array([1,2,3,4])
minimum = np.min(s)
maximum = np.max(s)
print("minimum:",minimum)
print("maximum:",maximum)
```

# 6.Random Sampling Functions

# 7.Boolean & Logical Functions

```python
# all() if all elements are True then true
val=np.array([1,2,3,0])
print(np.all(val))

val1 = np.array([1,2,3])
print(np.all(val1))
```

```python
# any() any if one element is True then true
val= np.array([1,2,3,0])
print(np.any(val))

vals= np.array([0,0,0,0])
print(np.any(vals))
```

# 8.Set Operations

```python
# Intersection of two arrays
a = np.array([1,2,3,4])
b = np.array([3,4,5,6])
intersection = np.intersect1d(a,b) # intersect1d here 1d means one dimension
print(intersection)
```

```python
# Union of two arrays
a = np.array([1,2,3,4])
b = np.array([3,4,5,6])
union = np.union1d(a,b)
print(union)
```

# 9.Array Attribute Functions

```python
a = np.array([1,2,3])
shape = a.shape # shape of array
size = a.size # number of elements
dimensions = a.ndim # number of dimensions
dtype = a.dtype # Data type of array

print("Shape of a:", shape)
print("Size of a:", size)
print("Number of dimensions of a:", dimensions)
print("Data type of a:", dtype)
```
