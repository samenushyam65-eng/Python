# Fortran and C Order

```python
import numpy as np 
num_array = np.arange(100).reshape(20,5)
print(num_array)

print(" ")

print("\nC Order (Iteration happens row by row, left to right)")    
for n in np.nditer(num_array,order="C"):
    print(n,end=" ")
    
print(" ")

print("\nF Order (Iteration happens column by column, top to bottom)")    
for n in np.nditer(num_array,order="F"):
    print(n,end=" ")
```

# Python Program to Generate OTP

```python
import random

def generate_otp(length=4):
    digits = '012345'
    otp = ''.join(random.choice(digits) for _ in range(length))
    return otp

# otp = generate_otp(4)
print(f"Your OTP is: {otp}")
```
