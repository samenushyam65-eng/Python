# Broadcasting: Used in Vectorization
The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

# Broadcasting Rules
1. Make the two arrays have the same number of dimensions.
2. Make each dimension of the two arrays the same size.

(3,1) + (1,4) = (3,4)  in first column i.e, (1) = in second row i.e, (1) braodcasting possible

# Broadcasting Rules
NumPy compares array shapes from right to left:
- If dimensions are equal → ✅ compatible.
- If one dimension is 1 → ✅ it can be stretched.
- If neither matches → ❌ error.
- The result shape is the maximum size in each dimension.

# docstring
A docstring is a special string in Python that documents a module, class, function, or method. It explains what the code does, how to use it, and sometimes gives examples.


```python
def add(a, b):
    """
    Add two numbers and return the result.

    Parameters:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    return a + b
```

```python
a=add(3,4)
print(a)
```

```python
print(add.__doc__) # to access the text written by us
```

```python
# Function Docstring
def greet(name):
    """Return a friendly greeting."""
    return f"Hello, {name}!"

print(greet.__doc__)
# Output: Return a friendly greeting.
```

```python
# class Docstring
class Dog:
    """A simple Dog class."""
    pass

print(Dog.__doc__)
# Output: A simple Dog class.
```

# np.histogram

```python
import numpy as np

data = [1, 2, 2, 3, 4, 5, 5, 5, 6]
hist, edges = np.histogram(data, bins=5)

print(hist)   # [1 2 1 1 4]
print(edges)  # [1. 2. 3. 4. 5. 6.]
```

- Bin [1,2] → contains 1 → count = 1
- Bin [2,3] → contains 2,2 → count = 2
- Bin [3,4] → contains 3 → count = 1
- Bin [4,5] → contains 4  → count = 1
- Bin [5,6] → contains 5,5,5,6 → count = 4     ← last bin includes the right edge

So the conversion is: raw data → bin edges → frequency counts

# 🧮 How It Converts Data
Correlation is derived from covariance normalization:
R_{ij}=\frac{C_{ij}}{\sqrt{C_{ii}\cdot C_{jj}}}
- C_{ij} = covariance between variable i and j.
- C_{ii}, C_{jj} = variances of each variable.
- This scaling ensures values are between -1 and 1:
- 1 → perfect positive correlation
- -1 → perfect negative correlation
- 0 → no linear correlation
