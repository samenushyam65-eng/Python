# Broadcasting: Used in Vectorization
The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

# Broadcasting Rules
1. Make the two arrays have the same number of dimensions.
2. Make each dimension of the two arrays the same size.

(3,1) + (1,4) = (3,4)  in first column i.e, (1) = in second row i.e, (1) braodcasting possible

# Broadcasting Rules
NumPy compares array shapes from right to left:
- If dimensions are equal → ✅ compatible.
- If one dimension is 1 → ✅ it can be stretched.
- If neither matches → ❌ error.
- The result shape is the maximum size in each dimension.

# docstring
A docstring is a special string in Python that documents a module, class, function, or method. It explains what the code does, how to use it, and sometimes gives examples.


```python
def add(a, b):
    """
    Add two numbers and return the result.

    Parameters:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    return a + b
```

```python
a=add(3,4)
print(a)
```

```python
print(add.__doc__) # to access the text written by us
```

```python
# Function Docstring
def greet(name):
    """Return a friendly greeting."""
    return f"Hello, {name}!"

print(greet.__doc__)
# Output: Return a friendly greeting.
```

```python
# class Docstring
class Dog:
    """A simple Dog class."""
    pass

print(Dog.__doc__)
# Output: A simple Dog class.
```
