The line %matplotlib inline is a Jupyter Notebook magic command. It tells Jupyter to display plots from Matplotlib directly inside the notebook cells, rather than opening them in a separate window.


```python
%matplotlib inline
x1 = np.linspace(0,10,50)

plt.plot(x1,np.sin(x1),'-')
plt.plot(x1,np.cos(x1),'--')
```

```python
# create the first of two panels and set current axis
plt.subplot(2,1,1) # (rows, columns, panel number)
plt.plot(x1,np.cos(x1),'*')
```

```python
# create a plot figure
plt.figure()

# create the first of two panels and set current axis
plt.subplot(2,1,1) # (rows, columns, this plot is the first plot)
plt.plot(x1,np.sin(x1))

# create the second of two panels and set current axis
plt.subplot(2, 1, 2)   # (rows, columns, this plot is the second plot)
plt.plot(x1, np.cos(x1));
```

# Object-Oriented API

```python
# First create a grid of plots
# ax will be an array of two Axes objects
fig, ax = plt.subplots(2,1)  # 2 rows ,1 column


# Call plot() method on the appropriate object
ax[0].plot(x1, np.sin(x1), 'b-')
ax[1].plot(x1, np.cos(x1), 'w-');
```

# Objects and Reference

```python
fig = plt.figure()

x2 = np.linspace(0, 5, 10)
y2 = x2 ** 2

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8])      # axes = fig.add_axes([left, bottom, width, height])

axes.plot(x2, y2, 'r')

axes.set_xlabel('x2')
axes.set_ylabel('y2')
axes.set_title('title');
```

# Error Bar Chart

```python
x9 = np.arange(0, 4, 0.2)

y9 = np.exp(-x9)

e1 = 0.1 * np.abs(np.random.randn(len(y9)))

plt.errorbar(x9, y9, yerr = e1, fmt = '.-')

plt.show();
```

# Stacked Bar Chart

```python
A = [15., 30., 45., 22.]

B = [15., 25., 50., 20.]

z1 = range(4)

plt.bar(z1,A,color='b')
plt.bar(z1,B,color='r',bottom=A)
plt.show()
```
