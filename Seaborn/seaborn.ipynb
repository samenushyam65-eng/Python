# FacetGrid

```python
# loading of a dataframe from seaborn
df = sns.load_dataset('tips')

############# Main Section         #############
custom_colors = {"Thur": "red", "Fri": "blue", "Sat": "green", "Sun": "purple"}
# Form a facetgrid using columns with a hue
graph = sns.FacetGrid(df, col ="sex",  hue ="day",height=5, aspect=1.5,palette=custom_colors)  
# height is the size of each facet, aspect is the ratio of width to height of each facet(subplot)

# map the above form facetgrid with some attributes
graph.map(plt.scatter, "total_bill", "tip", edgecolor ="w",alpha=0.5).add_legend()

# plt.setp(graph._legend.get_texts(), fontsize=12)   # legend labels

# show the object
plt.show()
```

# PairGrid

```python
# loading dataset
df = sns.load_dataset('tips')

custom_colors = {"Thur": "red", "Fri": "blue", "Sat": "green", "Sun": "purple"}
# PairGrid object with hue
graph = sns.PairGrid(df, hue ='day', height=5, aspect=1.5,palette=custom_colors)  
# height is the size of each facet, aspect is the ratio of width to height of each facet(subplot)
# type of graph for diagonal
graph = graph.map_diag(plt.hist)
# type of graph for non-diagonal
graph = graph.map_offdiag(plt.scatter)
# to add legends
graph = graph.add_legend()
# to show
plt.show()
```

```python
# loading dataset
df = sns.load_dataset('tips')

# PairGrid object with hue
graph = sns.PairGrid(df,height=5, aspect=1.5)
# type of graph for non-diagonal(upper part)
graph = graph.map_upper(sns.scatterplot)
# type of graph for non-diagonal(lower part)
graph = graph.map_lower(sns.kdeplot)
# type of graph for diagonal
graph = graph.map_diag(sns.kdeplot, lw = 2)
# to show
plt.show()
```

# Relational Plots



To draw the relational plots seaborn provides three functions. These are:

- relplot()
- scatterplot()
- lineplot()

Seaborn.relplot()

Example 1: Visualizing the most basic plot to show all the data points in tips dataset.

```python
# selecting the style
sns.set_style='ticks'

# reading the datasets
tips = sns.load_dataset('tips')

# plotting a simple visualization of data points
sns.relplot(x='total_bill',y='tip',data =tips,height=5,aspect=1.5)
```

```python
import matplotlib.colors as mcolors
data = np.random.rand(10, 10) * 100  

plt.figure(figsize=(12, 8))  # Adjust figure size

sns.heatmap(
    data, 
    xticklabels=list("ABCDEFGHIJ"),  # Custom x-axis labels
    yticklabels=False,  # Hide y-axis labels
    norm=mcolors.LogNorm(),  # Logarithmic scaling
    cmap="coolwarm",  # Color map
    linewidths=0.5  # Cell spacing
)

# Add title and labels
plt.title("Custom Heatmap", fontsize=16)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

Explanation:

- plt.figure(figsize=(12, 8)) enlarges the figure for better visibility.
- xticklabels=list("ABCDEFGHIJ") replaces default labels with A-J.
- yticklabels=False removes y-axis labels for clarity.
- norm=mcolors.LogNorm() enhances contrast for varied data.
- cmap="coolwarm" uses a blue-to-red gradient.
- linewidths=0.5 adds thin separation lines.
- plt.title("Custom Heatmap", fontsize=16) adds a title with a readable font size.
- plt.xlabel("X-axis") & plt.ylabel("Y-axis") labels the axes for clear understanding.
